```python
# Adding a new column to the census DataFrame to calculate the percentage of Muslims in the population 

census['Muslim_Pct'] = (census['Muslim_Males'] + census['Muslim_Females']) / census['Total_Population'] * 100
```

```python
# Filtering the census DataFrame to include only rows where the 'Level' column is 'Tehsil'

tehsil_df = census[census['Level'] == 'Tehsil']
```

```python
# Shape of the filtered DataFrame

tehsil_df.shape
```

```python
# Verified Muslim_Pct for Gurdaspur tehsil against known Radcliffe-era figure (52.16%) - matched (52.155%), OCR looks solid for this row

tehsil_df[(tehsil_df['District']=='Gurdaspur') & (tehsil_df['Tehsil']=='Gurdaspur')]
```

```python
# Storing all known anomalous values in a list of dictionaries for further analysis and comparison with the calculated Muslim_Pct values from the census DataFrame
known_values = [
    {'District': 'Gurdaspur', 'Tehsil': 'Gurdaspur', 'Expected_Pct': 52.16},
    {'District': 'Gurdaspur', 'Tehsil': 'Batala', 'Expected_Pct': 55.07},
    {'District': 'Amritsar', 'Tehsil': 'Ajnala', 'Expected_Pct': 59.46},
    {'District': 'Ferozepore', 'Tehsil': 'Ferozepore', 'Expected_Pct': 55.25},
    {'District': 'Ferozepore', 'Tehsil': 'Zira', 'Expected_Pct': 65.26},
    {'District': 'Jullundur', 'Tehsil': 'Jullundur', 'Expected_Pct': 51.16},
    {'District': 'Jullundur', 'Tehsil': 'Nakodar', 'Expected_Pct': 59.41},
]
```

```python
# Comparing known anomalous values with calculated Muslim_Pct values
for item in known_values:
    row = tehsil_df[(tehsil_df['District'] == item['District']) & (tehsil_df['Tehsil'] == item['Tehsil'])]
    actual = row['Muslim_Pct'].values[0]
    print(item['District'], item['Tehsil'], 'Expected:', item['Expected_Pct'], 'Actual:', round(actual, 2))
```

```python
# Calculating the sum of Total_Population for each District in the tehsil_df DataFrame

tehsil_sums = tehsil_df.groupby('District')['Total_Population'].sum()
tehsil_sums
```

```python
# Creating a new DataFrame to store the sum of Total_Population for each District

district_rows = census[census['Level'] == 'District'][['District', 'Total_Population']]
district_rows
```

```python
# Merging the district_rows DataFrame with the tehsil_sums Series to compare the Total_Population values

comparison = district_rows.merge(tehsil_sums, on='District')
comparison
```

## Resolving OCR Errors Manually

District totals were cross-checked against summed tehsil values — 14 mismatches found. 
Each will be verified against the original 1941 Census PDF before correcting.

```python
# Correcting OCR errors in the census DataFrame for Multan District
 
census.loc[(census['District'] == 'Multan') & (census['Tehsil'] == 'Multan') & (census['Level'] == 'District'), 'Muslim_Males'] = 629762
census.loc[(census['District'] == 'Multan') & (census['Tehsil'] == 'Multan') & (census['Level'] == 'District'), 'Total_Population'] = 1484333
```

```python
#checking the corrected values for Multan District

census.loc[(census['District'] == 'Multan')]
```

```python
# Correcting OCR errors in the census DataFrame for Ferozepore District

census.loc[(census['District'] == 'Ferozepore') & (census['Level'] == 'District'), 'Total_Population'] = 1422076
```

```python
#checking the corrected values for Ferozepore District

census.loc[(census['District'] == 'Ferozepore')]
```

### Shahpur — Investigated, No OCR Error Found

District Total_Population (998,921) is internally consistent with the sum of its own listed demographic columns, and every demographic column matches cleanly against the sum of its 4 tehsils (Shahpur, Khushab, Bhalwal, Sargodha) — checked and re-verified column by column.

The 1,001 gap exists only between the two *standalone* Total_Population figures (district vs. sum of tehsil totals) and was manually confirmed against the original PDF — not an OCR misread. Likely reflects a rounding or compilation discrepancy in the original 1941 table itself.

**Conclusion:** left as-is, no correction applied. Logged as a genuine data limitation rather than a fixable error.
