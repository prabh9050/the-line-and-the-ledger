## Data quality note: 1951 East Punjab religion counts

In several districts, `Total Population` is larger than the sum of the religion columns.
The 268,602 people with no religion breakdown are mostly in Amritsar (96,720),
Gurdaspur (89,512) and Jalandhar (46,834).

Cause: a fire at the Jullundur census tabulation office destroyed the records


**Handling:** shares are computed from `rel_total` (sum of the religion columns),
not `Total Population`, so each district's shares add up to 100%.
The `gap` column keeps the missing counts visible.

```python
rel_cols = ['Hinduism','Sikhism','Islam','Christianity','Jainism','Others']

east_census_1951['rel_total'] = east_census_1951[rel_cols].sum(axis=1)
east_census_1951['gap'] = east_census_1951['Total Population'] - east_census_1951['rel_total']

east_census_1951[['District','rel_total','Total Population','gap']].sort_values('gap', ascending=False)
```

```python
name_map = {'Jullundur':'Jalandhar', 'Ferozepore':'Firozpur',
            'Layallpur':'Lyallpur', 'Attock':'Campbellpur'}
census_1941['District'] = census_1941['District'].replace(name_map)

rel_cols = ['Hinduism','Sikhism','Islam','Christianity','Jainism','Others']
census_1941['rel_total'] = census_1941[rel_cols].sum(axis=1)
census_1941['gap'] = census_1941['Total Population'] - census_1941['rel_total']
print(census_1941['gap'].abs().max())
```

```python
west_census_1951 = west_census_1951.rename(columns={
    'Islam_Pop':'Islam', 'Christianity_Pop':'Christianity',
    'Hinduism_Pop':'Hinduism', 'Total_Pop':'Total Population'})

west_census_1951['District'] = west_census_1951['District'].replace(
    {'Campbellpur (Attock)':'Campbellpur'})

west_census_1951['Sikhism'] = np.nan

west_census_1951['Others'] = (west_census_1951['Others_Pop']
    + west_census_1951['Zoroastrianism_Pop'] + west_census_1951['Buddhism_Pop'])
west_census_1951['Jainism'] = np.nan
```

```python
west_census_1951['rel_total'] = west_census_1951[rel_cols].sum(axis=1)
west_census_1951['gap'] = west_census_1951['Total Population'] - west_census_1951['rel_total']
```

```python
rename_map = {'Hinduism':'Hindus', 'Sikhism':'Sikhs', 'Islam':'Muslims'}

for df in [census_1941, east_census_1951, west_census_1951]:
    df.rename(columns=rename_map, inplace=True)

rel_cols = ['Hindus','Sikhs','Muslims','Christianity','Jainism','Others']
```

## Data prep: renaming, merging assumptions, and missing data

**What we did**
- Renamed `Hinduism`, `Sikhism`, `Islam` to `Hindus`, `Sikhs`, `Muslims` in all three tables.
- Standardised district names across years (Jullundur -> Jalandhar, Ferozepore -> Firozpur, Layallpur -> Lyallpur, Attock -> Campbellpur).
- Added `rel_total` (sum of religion columns) and `gap` (reported total minus `rel_total`).
  Shares use `rel_total` so they add up to 100%.

**Missing / unknown data**
- East Punjab 1951: 268,602 people have no religion breakdown (gap), mostly in
  Amritsar, Gurdaspur and Jalandhar. Cause: fire at the Jullundur census office
  
- West Punjab 1951 has no Sikh column. Set to `NaN` (unknown), not 0, so we
  don't claim Sikhs were absent.

- West Punjab 1951: merged `Zoroastrianism_Pop` + `Buddhism_Pop` + `Others_Pop` into
  `Others` (tiny counts) so columns match the other tables. No Jainism column,
  so `Jainism` is `NaN` (unknown), same as Sikhs.
- Added asserts: 1941 and West gaps are exactly 0, East total gap is 268,602.

**Assumptions USED in calculations**
- Scheduled Castes are assumed to be included in `Hindus` in the 1951 tables
  If wrong, 1951 Hindu counts would be undercounted and the 1941-to-1951
  drop overstated. West Hindu counts are tiny (e.g. Lahore 2,433), so the
  overall pattern would hold, only the size of the drop would change.

**Assumptions NOT used (just noted)**
- Sikhs may have been counted under Hindus in Pakistan's 1951 tables.
  Unconfirmed, so we do not merge them. West Sikhs are left as `NaN`.

```python
# calculate each group's share (percentage) in every table
for table in [census_1941, east_census_1951, west_census_1951]:
    for group in ['Hindus', 'Sikhs', 'Muslims']:
        table[group + '_share'] = table[group] / table['rel_total'] * 100

# keep only the columns we need, and stack East and West for 1951
share_columns = ['District', 'Hindus_share', 'Sikhs_share', 'Muslims_share']

shares_1941 = census_1941[share_columns].copy()
shares_1951 = pd.concat([east_census_1951[share_columns],west_census_1951[share_columns]])

# put 1941 and 1951 side by side, matching rows by district name
shift = shares_1941.merge(shares_1951, on='District', suffixes=('_1941', '_1951'))

# change in percentage points for all three groups
for group in ['Hindus', 'Sikhs', 'Muslims']:
    shift[group + '_change'] = shift[group + '_share_1951'] - shift[group + '_share_1941']

print(len(shift)) 
shift
```

```python
# label each district as East or West Punjab
east_districts = list(east_census_1951['District'])
shift['side'] = np.where(shift['District'].isin(east_districts),
                         'East Punjab', 'West Punjab')
is_east = shift['side'] == 'East Punjab'

# West Punjab Sikh share in 1951 is unknown (NaN), so use 0 for this chart only
sikh_share_1951_filled = shift['Sikhs_share_1951'].fillna(0)

# minority share for each district
west_minority_1941 = shift['Hindus_share_1941'] + shift['Sikhs_share_1941']
west_minority_1951 = shift['Hindus_share_1951'] + sikh_share_1951_filled

shift['minority_1941'] = np.where(is_east, shift['Muslims_share_1941'], west_minority_1941)
shift['minority_1951'] = np.where(is_east, shift['Muslims_share_1951'], west_minority_1951)

# draw two panels side by side
fig, axes = plt.subplots(1, 2, figsize=(14, 9), sharex=True)
panel_titles = {'East Punjab': 'East Punjab: Muslim share of population',
                'West Punjab': 'West Punjab: Hindu and Sikh share of population'}

for axis, side in zip(axes, ['East Punjab', 'West Punjab']):
    part = shift[shift['side'] == side].sort_values('minority_1941')
    positions = range(len(part))
    axis.hlines(positions, part['minority_1951'], part['minority_1941'], color='grey')
    axis.scatter(part['minority_1941'], positions, color='tab:blue', label='1941', zorder=3)
    axis.scatter(part['minority_1951'], positions, color='tab:red', label='1951', zorder=3)
    axis.set_yticks(list(positions))
    axis.set_yticklabels(part['District'])
    axis.set_title(panel_titles[side])
    axis.set_xlabel('Share of population (percent)')
    axis.legend()

fig.text(0.01, 0.01,
         'Notes: West Punjab 1951 Sikh counts are not reported, so they are treated as zero here. '
         'Christians are excluded from both panels.\n'
         'Gurdaspur and Sialkot are adjusted for Shakargarh tehsil; Lahore 1941 includes the part of Kasur that was split.\n'
         'Migrants from outside these districts (princely states, Delhi, Alwar, Uttar Pradesh, Bahawalpur) are in the 1951 counts.',
         fontsize=8)
plt.tight_layout(rect=[0, 0.04, 1, 1])
plt.savefig(reports_folder / 'minority_share_dumbbell.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# label every 1941 district as East or West, and add Hindus + Sikhs together
east_names = list(east_census_1951['District'])
counts_1941 = census_1941.copy()
counts_1941['side'] = np.where(counts_1941['District'].isin(east_names),
                               'East Punjab', 'West Punjab')
counts_1941['Hindus_and_Sikhs'] = counts_1941['Hindus'] + counts_1941['Sikhs']

# 1951 tables, labelled the same way (West Sikh count unknown, so 0 here)
east_1951 = east_census_1951.copy()
east_1951['side'] = 'East Punjab'
west_1951 = west_census_1951.copy()
west_1951['side'] = 'West Punjab'
west_1951['Sikhs'] = west_1951['Sikhs'].fillna(0)

counts_1951 = pd.concat([east_1951, west_1951])
counts_1951['Hindus_and_Sikhs'] = counts_1951['Hindus'] + counts_1951['Sikhs']

# add up each side, then subtract 1941 from 1951
groups = ['Muslims', 'Hindus_and_Sikhs']
total_1941 = counts_1941.groupby('side')[groups].sum()
total_1951 = counts_1951.groupby('side')[groups].sum()
net_change = total_1951 - total_1941
print(net_change)

# bar chart in millions of people
(net_change / 1_000_000).plot(kind='bar', figsize=(8, 6), rot=0)
plt.axhline(0, color='black', linewidth=0.8)
plt.ylabel('Change in population, 1941 to 1951 (millions)')
plt.title('Net change by community and side of the border')
fig.text(0.01, 0.01,
         'Notes: West Punjab 1951 Sikh counts are not reported, so they are treated as zero here. '
         'Christians are excluded from both panels.\n'
         'Gurdaspur and Sialkot are adjusted for Shakargarh tehsil; Lahore 1941 includes the part of Kasur that was split.\n'
         'Migrants from outside these districts (princely states, Delhi, Alwar, Uttar Pradesh, Bahawalpur) are in the 1951 counts.',
         fontsize=8)
plt.tight_layout(rect=[0, 0.08, 1, 1])
plt.savefig(reports_folder / 'net_change_by_community.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
from matplotlib.patches import Patch

east_names = list(east_census_1951['District'])

# concatenate 1941 and 1951 population counts, then calculate growth percent and side
population_1941 = census_1941[['District', 'Total Population']].rename(
    columns={'Total Population': 'population_1941'})
population_1951 = pd.concat([east_census_1951[['District', 'Total Population']],
                             west_census_1951[['District', 'Total Population']]]).rename(
    columns={'Total Population': 'population_1951'})

growth = population_1941.merge(population_1951, on='District')

# Calculate growth percent and side
growth['growth_percent'] = ((growth['population_1951'] - growth['population_1941'])
                            / growth['population_1941'] * 100)
growth['side'] = np.where(growth['District'].isin(east_names), 'East Punjab', 'West Punjab')

# marking districts that straddle the border with an asterisk
boundary_districts = ['Lahore']
growth['label'] = np.where(growth['District'].isin(boundary_districts),
                           growth['District'] + ' *', growth['District'])

# visualize growth percent by district, with East and West Punjab in different colors
growth = growth.sort_values('growth_percent')
bar_colors = np.where(growth['side'] == 'East Punjab', 'tab:orange', 'tab:green')

plt.figure(figsize=(8, 9))
plt.barh(growth['label'], growth['growth_percent'], color=bar_colors)
plt.axvline(0, color='black', linewidth=0.8)
plt.xlabel('Population change, 1941 to 1951 (percent)')
plt.title('Population growth by district')
plt.legend(handles=[Patch(color='tab:orange', label='East Punjab'),
                    Patch(color='tab:green', label='West Punjab')])
plt.figtext(0.01, 0.01,
            '* Lahore 1941 includes the part of Kasur that was split. Gurdaspur and Sialkot are adjusted for Shakargarh.\n'
            'Growth includes births, deaths and migration, including migrants from outside these districts.',
            fontsize=7)
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig(reports_folder / 'population_growth_by_district.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# calculate the share of each religious group in the total population for each side and year
groups = ['Hindus', 'Sikhs', 'Muslims']

def side_composition(table, label):
    totals = table[groups].sum()
    totals['Others'] = table[['Christianity', 'Jainism', 'Others']].sum().sum()
    shares = totals / totals.sum() * 100
    shares.name = label
    return shares

east_1941 = census_1941[census_1941['District'].isin(east_names)]
west_1941 = census_1941[~census_1941['District'].isin(east_names)]

composition = pd.DataFrame([
    side_composition(east_1941, 'East Punjab 1941'),
    side_composition(east_census_1951, 'East Punjab 1951'),
    side_composition(west_1941, 'West Punjab 1941'),
    side_composition(west_census_1951, 'West Punjab 1951'),
])
print(composition.round(1))

# plot the religious composition for each side and year
composition.plot(kind='bar', stacked=True, figsize=(9, 6), rot=0,
                 color=['tab:orange', 'tab:blue', 'tab:green', 'lightgrey'])
plt.ylabel('Share of population with known religion (percent)')
plt.title('Religious composition before and after, by side of the border')
plt.figtext(0.01, 0.01,
            'West 1951 Sikh counts are not reported (shown as zero). "Others" includes Christians, Jains and small groups.\n'
            'Sides assigned by district; Shakargarh moved to Sialkot; Lahore straddles the border.\n'
            'Both sides include migrants from outside these districts (princely states, Delhi, Alwar, Uttar Pradesh, Bahawalpur).',
            fontsize=7)
plt.tight_layout(rect=[0, 0.07, 1, 1])
plt.savefig(reports_folder / 'religious_composition_by_side.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Findings: 1941 vs 1951 district-level demographic shift

**What the data shows**
- **East Punjab (India side):** the Muslim share of the population fell from about a third in 1941 to 1.9% in 1951. The Hindu share rose to 66.1% and the Sikh share to 31.0%.
- **West Punjab (Pakistan side):** the Muslim share rose from about three quarters to 97.8%. Hindus and Sikhs together fell from about 23% to about 0.1% of the counted population. No West district kept more than about 1% Hindus.
- **Net change in head counts (district-level sums, Shakargarh moved to Sialkot):** East Punjab lost roughly 4 million Muslims and gained roughly 3.8 million Hindus and Sikhs. West Punjab lost roughly 3.6 million Hindus and Sikhs and gained roughly 6.5 million Muslims. Exact values are in the printed `net_change` table.
- **The one large partial exception:** Gurgaon kept about 17% Muslims in 1951 (down from 33.6%). No West district shows anything comparable. Malerkotla, a princely state outside this dataset, is described by Ishtiaq Ahmed as the sole exception to the pattern of minorities being driven out of Punjab.
- **Fastest growth:** the canal-colony districts of West Punjab, Lyallpur (+54%), Multan (+42%) and Montgomery (+37%). This is consistent with, but does not prove, irrigated land and settlement changing hands.

**What this does NOT show**
- These are changes in census counts, not death tolls. They mix natural growth, deaths, migration and boundary changes. This is why the Muslim gain in West Punjab is larger than the Muslim loss in East Punjab.
- Reliable district-level death numbers do not exist. Province-wide estimates range widely, and the human cost is better measured here through displacement than through deaths.
- Both sides received migrants from outside the districts in this table. A 1948 West Punjab refugee census (via Elisabetta Iob; figures widely regarded as unreliable) counts roughly 5.5 million refugees, about three quarters from East Punjab districts and the rest from East Punjab princely states (about 682,000), Jammu and Kashmir, Alwar, Delhi, Uttar Pradesh and elsewhere. East Punjab also received Hindu and Sikh refugees from outside the 16 West districts, for example from Bahawalpur state. These movements add to one side's gains without subtracting from the other side's losses in this table, and are part of why the Muslim gain in West Punjab is larger than the Muslim loss in East Punjab.

**Limits of this analysis**
- **Missing East 1951 data:** 268,602 people (mostly in Amritsar, Gurdaspur and Jalandhar) have no religion breakdown. This is attributed to a fire at the Jullundur census tabulation office, a cause verified by the project author. Shares are therefore calculated on `rel_total`, the people with a known religion.
- **West 1951 Sikhs and Jains are not reported.** They are stored as `NaN` (unknown). For charts they are shown as zero, which is an assumption, not a finding.
- **Scheduled Castes are assumed to be counted under Hindus in 1951.** This is unconfirmed. If wrong, the Hindu drop in West Punjab is overstated in size, but West Hindu counts are so small that the overall pattern would hold.
- **Sikhs may have been counted under Hindus in Pakistan's 1951 tables.** Unconfirmed, so it was NOT used in any calculation.
- **Boundary adjustment:** Shakargarh tehsil is moved from Gurdaspur to Sialkot in the 1941 data to match 1951 boundaries (see the case note in notebook 01). Lahore 1941 still includes the part of Kasur that went to India; that split happened below tehsil level and cannot be corrected with census tables. Districts are assigned to East or West as whole units, so figures differ slightly from the tehsil-level analysis.
- **Princely states are not in the data.** Patiala, Faridkot and other states are outside the 1941 British-district tables, but many of their people moved in 1947. Their movements are not captured here.
- **Christians and other small groups** are excluded from the minority-share charts, which show the community that left each side. They are included in the "Others" group in the stacked bar chart.
- The 1941 and 1951 tables come from secondary web sources, not the original census volumes.
